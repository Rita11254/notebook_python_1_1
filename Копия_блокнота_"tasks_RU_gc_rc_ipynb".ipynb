**Задача 1:** Изучите как работает базовый reference counting

Создайте строку и исследуйте ее reference count на разных этапах. Верните кортеж: (count_after_creation, count_after_ref, count_after_del)

```python
import sys
import gc
import weakref
```

```python
# когда короткая строка (неизменяемая)
def task1():
    s = "a"
    count_after_creation = sys.getrefcount(s) - 1

    s2 = s
    count_after_ref = sys.getrefcount(s) - 1

    del s2
    count_after_del = sys.getrefcount(s) - 1

    return (count_after_creation, count_after_ref, count_after_del)

# Проверка
print("Задача 1:", task1())
```

**Задача 3:** Циклические ссылки и Memory Leaks

Создайте два объекта с циклической ссылкой, затем разорвите ее.
Используйте gc для проверки количества собранных объектов.
Верните количество объектов собранных gc после разрыва ссылки.

```python
import sys
import gc
import weakref

def task3():
    class Node:
        def __init__(self, name):
            self.name = name
            self.ref = None

    gc.enable()
    gc.collect()

    a = Node("I")
    b = Node("You")
    a.ref = b
    b.ref = a

    a.ref = None
    b.ref = None

    del a
    del b

    posle_count = gc.collect()

    return posle_count

print("Задача 3:", task3())
```

**Задача 5:** Weak References

Создайте два объекта со слабой ссылкой (weakref) между ними.
Убедитесь, что объекты могут быть удалены сборщиком мусора.
Верните True если weakref не увеличивает reference count.

```python
import sys
import gc
import weakref

def task5():
    class Data:
        def __init__(self, value):
            self.value = value

    gc.collect()

    a1 = Data("I")
    a2 = Data("You")

    count_a1_after = sys.getrefcount(a1) - 1
    count_a2_after = sys.getrefcount(a2) - 1

    a1.weakref = weakref.ref(a2)
    a2.weakref = weakref.ref(a1)

    count_a1 = sys.getrefcount(a1) - 1
    count_a2 = sys.getrefcount(a2) - 1

    id_a1 = id(a1)
    id_a2 = id(a2)

    del a1
    del a2

    gc.collect()

    return bool((count_a1 and count_a1_after) or (count_a2 and count_a2_after))

print("Задача 5:", task5())
```

**Задача 6:** Мониторинг работы Garbage Collector

Зарегистрируйте callback функцию для отслеживания работы GC.
Верните список событий, которые отследил callback.

```python
import sys
import gc
import weakref

def task6():
  point = []

  def gc_callback(moment, info):
    # if moment == 'start':
    #   point.append("GC начал работу")
    # elif moment == 'stop':
    #   point.append("GC закончил работу")
    if moment == "start":
      point.append(f"GC начал работу: {info}")
    elif moment == "stop":
      point.append(f"GC закончил работу: {info}")
    elif moment == "collection":
      generation = info.get("generation")
      collected = info.get("collected", 0)
      uncollectable = info.get("uncollectable", 0)
      point.append(f"Поколение: {generation} собрано объектов: {collected}, не собранно объектов: {uncollectable}")

  gc.callbacks.append(gc_callback)

  x = [i for i in range(2025)]
  del x
  gc.collect()
  gc.callbacks.remove(gc_callback)

  return point

print("Задача 6:", task6())
```

**Задача 7:** Анализ поколений в GC

Создайте объекты и отследите их перемещение между поколениями GC.
Верните кортеж с количеством объектов в каждом поколении до и после создания объектов.

```python
import sys
import gc
import weakref

def task7():
    gc.enable()
    gc.collect()

    class point:
        def __init__(self, id):
            self.id = id

    start = gc.get_count()
    print(f"Начальное состояние: gen_0 = {start[0]}, gen_1 = {start[1]}, gen_2 = {start[2]}")

    temp = [point(j) for j in range(2025)]
    initial = gc.get_count()
    print(f"После создания объекта: gen_0 = {initial[0]}, gen_1 = {initial[1]}, gen_2 = {initial[2]}")

    gc.collect(0)
    after_gen0_collect = gc.get_count()
    print(f"Почистив 0ое поколение: gen_0 = {after_gen0_collect[0]}, gen_1 = {after_gen0_collect[1]}, gen_2 = {after_gen0_collect[2]}")

    gc.collect(1)
    after_gen1_collect = gc.get_count()
    print(f"Почистив 1ое поколение: gen_0 = {after_gen1_collect[0]}, gen_1 = {after_gen1_collect[1]}, gen_2 = {after_gen1_collect[2]}")

    gc.collect(2)
    after_gen2_collect = gc.get_count()
    print(f"Почистив 2ое поколение: gen_0 = {after_gen2_collect[0]}, gen_1 = {after_gen2_collect[1]}, gen_2 = {after_gen2_collect[2]}")

    del temp
    gc.collect()

    final_state = gc.get_count()
    print(f"После всех удалений: gen_0 = {final_state[0]}, gen_1 = {final_state[1]}, gen_2 = {final_state[2]}")

    return (start, initial, after_gen0_collect, after_gen1_collect, after_gen2_collect, final_state)

print("Задача 7:", task7())
```

**Задача 9:** Быстрая проверка работы GC

Проверьте, собирает ли GC циклические ссылки.
Верните разницу в количестве объектов до и после сборки.

```python
import sys
import gc
import weakref

def task9():
    gc.disable()

    a, b, c = [], [], []
    a.append(b)
    b.append(c)
    c.append(a)

    count_before = len(gc.get_objects())

    collected = gc.collect()

    count_after = len(gc.get_objects())

    return count_before - count_after, collected

print("Задача 9:", task9())
```

```python
def task9_solution():
    # Создаем циклические ссылки быстро
    a, b, c = [], [], []
    a.append(b); b.append(c); c.append(a)  # Цикл a->b->c->a

    count_before = len(gc.get_objects())
    collected = gc.collect()
    count_after = len(gc.get_objects())

    return count_before - count_after, collected

print("Задача 9:", task9_solution())
```

**Задача 10:** Детектор "неубиваемых" объектов

Найдите объекты, которые переживают принудительную сборку мусора.
Выведите количество таких объектов.

```python
import sys
import gc
import weakref

def task10():
    gc.set_debug(gc.DEBUG_SAVEALL)

    class Persistent:
        def __del__(self):
            pass

    obj_1 = Persistent()
    obj_2 = Persistent()

    obj_1.ref = obj_2
    obj_2.ref = obj_1

    del obj_1, obj_2
    gc.collect()

    result = len(gc.garbage)

    gc.garbage.clear()
    gc.set_debug(0)

    itog = len(gc.garbage)

    return result, itog
print("Задача 10:", task10())
```
